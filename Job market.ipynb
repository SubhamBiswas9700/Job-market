```python
import pandas as pd
```

```python
# Loaded Excel file 
df = pd.read_excel("ai_job_market_insights.xlsx")
```

```python
# Display basic info
print(df.shape)
df.head()
```

```python
# Clean column names
df.columns = df.columns.str.lower().str.strip()
```

```python
# Check for missing values
print(df.isnull().sum())
```

```python
df.columns = df.columns.str.lower().str.strip()
```

```python
print(df.columns.tolist())
```

```python
df['job_title'] = df['job_title'].str.lower().str.strip()
```

```python
# Clean skills (remove spaces, lower case, split into list)
df['required_skills'] = df['required_skills'].str.lower().str.replace(' ', '')
df['skills list'] = df['required_skills'].str.split(',')
```

```python
# Clean location (we’ll extract city later)
df['location'] = df['location'].str.strip()
```

```python
# Convert salary to numeric
df['salary_usd'] = pd.to_numeric(df['salary_usd'], errors='coerce')
```

```python
Average Salary by Industry
```

```python
industry_salary = df.groupby('industry')['salary_usd'].mean().sort_values(ascending=False)
print(industry_salary.head(10))
```

```python
Average Salary by Job Title
```

```python
title_salary = df.groupby('job_title')['salary_usd'].mean().sort_values(ascending=False)
print(title_salary.head(10))
```

```python
Number of Jobs by Industry
```

```python
job_count_industry = df['industry'].value_counts()
print(job_count_industry.head(10))
```

```python
 Remote vs Onsite Job Count
```

```python
remote_jobs = df['remote_friendly'].value_counts()
print(remote_jobs)
```

```python
High Salary Jobs
```

```python
high_salary_jobs = df[df['salary_usd'] > 100000]  # Change threshold as needed
print(high_salary_jobs[['job_title', 'industry', 'salary_usd']].head())
```
